```python
from flask import Flask, render_template, request
import nltk
from nltk import Tree
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('maxent_ne_chunker')
nltk.download('words')

app = Flask(__name__)

@app.route('/')
def home():
    message = "Welcome to Week 11"
    return render_template('index.html', message=message)

@app.route('/greet', methods=['POST'])
def greet():
    user_name = request.form['user_name']
    greeting_message = f"Hello, {user_name}! Welcome to Week 11. Tell me a bit about yourself."
    return render_template('index.html', message=greeting_message)

@app.route('/analyze', methods=['POST'])
def analyze():
    # Get user input text for analysis
    user_text = request.form['user_text']
    
    # Tokenize the user input text
    tokens = word_tokenize(user_text)
    
    # Basic preprocessing
    tokens = [token.lower() for token in tokens if token.isalpha()]
    tokens = [token for token in tokens if token not in stopwords.words('english')]
    
    # Frequency distribution analysis
    fdist = FreqDist(tokens)
    most_common = fdist.most_common(5)
    
    # Part-of-speech tagging
    tagged_tokens = nltk.pos_tag(tokens[:20])
    
    # Named entity recognition
    entities = nltk.ne_chunk(tagged_tokens)

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens[:20]]

    # Render the results to the user
    analysis_results = {
        "most_common": most_common,
        "tagged_tokens": tagged_tokens,
        "lemmatized_tokens": lemmatized_tokens,
        "entities": entities
    }
    return render_template('index.html', message="Text analysis complete!", analysis_results=analysis_results)

if __name__ == "__main__":
    app.run(host='0.0.0.0', port=5000, debug=False)
```
